```python
%matplotlib inline
from qiskit_aer import Aer
from qiskit import QuantumCircuit
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt
import numpy as np
import math
```

```python
n = 3
N = 2**n
qc = QuantumCircuit(n, n)
qc.h(range(n))

# Step 2: Oracle (marking the target state |11...1|)
oracle = QuantumCircuit(n)
oracle.cz(0, 1)
qc.compose(oracle, inplace=True)
```

```python
# Diffuser (inversion about the mean)
def diffuser(n):
    diff = QuantumCircuit(n)
    diff.h(range(n))
    diff.x(range(n))
    diff.h(n - 1)
    diff.mcx(list(range(n - 1)), n - 1)
    diff.h(n - 1)
    diff.x(range(n))
    diff.h(range(n))
    return diff
```

```python

R = math.floor((math.pi / 4) * math.sqrt(N))

# Apply the oracle + diffuser R times
for _ in range(R):
    qc.append(oracle, range(n))  # Apply oracle
    qc.append(diffuser(n), range(n))  # Apply diffuser

# Step 5: Measurement
qc.measure(range(n), range(n))

# Step 6: Simulation
simulator = Aer.get_backend('qasm_simulator')
job = simulator.run(qc, shots=1024)
result = job.result()
counts = result.get_counts()

# Output the results
print("Measurement Counts:", counts)
plot_histogram(counts)
plt.show()
```
