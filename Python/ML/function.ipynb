```python
import google.generativeai as genai
import random
```

```python
def get_temperature(location):
    temp = random.randint(-10, 45)
    return f"The temperature in {location} is {temp}°C."


def get_humidity(location):
    humidity = random.randint(20, 100)
    return f"The humidity in {location} is {humidity}%."


def get_wind_speed(location):
    wind = random.uniform(1, 20)
    return f"The wind speed in {location} is {wind:.1f} km/h."


def get_uv_index(location):
    uv = random.randint(0, 11)
    return f"The UV index in {location} is {uv}."


def get_air_quality(location):
    aqi = random.randint(0, 500)
    return f"The air quality index in {location} is {aqi}."


def get_weather_condition(location):
    conditions = ["sunny", "cloudy", "rainy",
                  "snowy", "foggy", "windy", "stormy"]
    condition = random.choice(conditions)
    return f"The weather condition in {location} is {condition}."


def get_precipitation_chance(location):
    chance = random.randint(0, 100)
    return f"The chance of precipitation in {location} is {chance}%."


def get_feels_like_temperature(location):
    feels_like = random.randint(-15, 50)
    return f"The 'feels like' temperature in {location} is {feels_like}°C."


def get_sunrise_time(location):
    hour = random.randint(5, 7)
    minute = random.randint(0, 59)
    return f"The sunrise time in {location} is {hour}:{minute:02d} AM."


def get_sunset_time(location):
    hour = random.randint(5, 8) + 12
    minute = random.randint(0, 59)
    return f"The sunset time in {location} is {hour}:{minute:02d} PM."
```

```python
# Configure your API key
genai.configure(api_key="API_KEY")


# Initialize the Gemini model
model = genai.GenerativeModel("gemini-2.0-flash-lite")

while True:
    # User asks a question
    user_input = input()
    if(not user_input): break
    print("User:", user_input)

    # Send the prompt to the model
    response = model.generate_content(f"""
    You are a helpful assistant. If a user asks for whether related info, use the following functions:
    def get_temperature(location)
    def get_humidity(location)
    def get_wind_speed(location)
    def get_uv_index(location)
    def get_air_quality(location)
    def get_weather_condition(location):
    def get_precipitation_chance(location):
    def get_feels_like_temperature(location):
    def get_sunrise_time(location)
    def get_sunset_time(location)

    If user asks for anything related to above function, just respond with corresponding function call
    "CALL relevant_function(location="city name")" only and nothing else. If the functions are not relevant
    to the prompt, reply to steer the coversation by saying the available options for the user
    Keep the replies to 80 characters per line

    Now, the user asked: "{user_input}"
    """)

    model_reply = response.text.strip()
    if model_reply.startswith("CALL"):
        start = model_reply.find('location="') + len('location="')
        end = model_reply.find('"', start)
        location = model_reply[start:end]
        function_response = eval(model_reply.split()[1])
        final_response = model.generate_content(f"""
    User asked: "{user_input}"
    Function relevant_function(location="{location}") was called.
    It returned: "{function_response}"

    Respond to the user appropriately. With a few line of logical explaination
    why this might be happening 80 chars per line
    """)
        print("Bot:", final_response.text.strip())
    else:
        print("Bot:", model_reply)
    print()
```
